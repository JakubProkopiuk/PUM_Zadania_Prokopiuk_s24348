```python

from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns

root = Path.cwd()
while not (root / "conf" / "base" / "catalog.yml").exists():
    if root == root.parent:
        raise RuntimeError("Nie znaleziono katalogu projektu Kedro")
    root = root.parent

folder = root / "data" / "08_reporting"
folder.mkdir(parents=True, exist_ok=True)

sns.set_theme(style="whitegrid")

def zapisz(nazwa):
    plt.tight_layout()
    plt.savefig(folder / f"{nazwa}.png", dpi=150)
    plt.show()
    plt.close()

# ===== TITANIC =====

plt.figure(figsize=(8, 5))
sns.histplot(data=titanic, x="age", bins=30)
plt.title("Titanic - rozklad wieku pasazerow")
zapisz("01_titanic_wiek")

plt.figure(figsize=(8, 5))
sns.countplot(data=titanic, x="survived")
plt.title("Titanic - liczba osob wedlug przezycia")
plt.xlabel("Przezycie: 0 = nie, 1 = tak")
zapisz("02_titanic_przezycie")

plt.figure(figsize=(8, 5))
sns.barplot(data=titanic, x="sex", y="survived", errorbar=None)
plt.title("Titanic - udzial ocalalych wedlug plci")
plt.ylabel("Odsetek ocalalych")
zapisz("03_titanic_plec")

# ===== IRIS =====

iris["species"] = iris["target"].map({
    0: "Setosa",
    1: "Versicolor",
    2: "Virginica"
})

plt.figure(figsize=(8, 5))
sns.histplot(data=iris, x="petal length (cm)", bins=20)
plt.title("Iris - rozklad dlugosci platkow")
zapisz("04_iris_platki")

plt.figure(figsize=(8, 5))
sns.countplot(data=iris, x="species")
plt.title("Iris - liczba kwiatow wedlug gatunku")
zapisz("05_iris_gatunki")

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=iris,
    x="petal length (cm)",
    y="petal width (cm)",
    hue="species"
)
plt.title("Iris - zaleznosc wymiarow platkow")
zapisz("06_iris_zaleznosc")

print("GOTOWE! Zapisano 6 wykresow.")
print("Lokalizacja:", folder)
```
